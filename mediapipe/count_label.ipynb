```python
from collections import Counter
import glob
import os

label_dir = "/home/kth/robot_dev/mediapipe/data/drivision2/valid/labels"
counter = Counter()

for path in glob.glob(os.path.join(label_dir, '*.txt')):
    with open(path, 'r') as f:
        for line in f:
            cls = int(line.strip().split()[0])
            counter[cls] += 1
print(counter)
```

```python
for path in glob.glob(os.path.join(label_dir, '*.txt')):
    with open(path, 'r') as f:
        for line in f:
            length = len(line.strip().split())
            counter[length] += 1
print(counter)
```
